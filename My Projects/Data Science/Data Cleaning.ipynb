```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import numpy as np
import os


file_path = "collected_data/NVDA_historical_data.csv"
df = pd.read_csv(file_path)

print("Dataset loaded for cleaning.")
print(f"Shape before cleaning: {df.shape}")
```

```python

missing_count = df.isnull().sum()
print("Missing values per column:\n", missing_count)


df.ffill(inplace=True)


df.dropna(inplace=True)

print(f"Shape after handling missing values: {df.shape}")
```

```python

numeric_df = df.select_dtypes(include=[np.number])


target_cols = ['Open', 'High', 'Low', 'Close', 'Volume']
available_numeric_cols = [col for col in target_cols if col in numeric_df.columns]

print(f"Columns selected for scaling: {available_numeric_cols}")


scaler = MinMaxScaler(feature_range=(0, 1))
df_scaled = df.copy()

if len(available_numeric_cols) > 0:

    df_scaled[available_numeric_cols] = scaler.fit_transform(df[available_numeric_cols])
    print("Success: Data normalized.")
else:
    print("Error: No numeric columns found to scale.")

print(df_scaled.head())
```

```python
processed_dir = "processed_data"
if not os.path.exists(processed_dir):
    os.makedirs(processed_dir)


clean_file_path = os.path.join(processed_dir, "NVDA_cleaned_data.csv")
df_scaled.to_csv(clean_file_path, index=False)

print(f"Cleaned dataset saved at: {clean_file_path}")
```
