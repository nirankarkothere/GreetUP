```python
import yfinance as yf
import pandas as pd
import os

print("Libraries imported successfully.")
```

```python
ticker_symbol = "NVDA"
start_date = "2024-01-01"
end_date = "2026-03-24"

try:
    print(f"Fetching data for {ticker_symbol}...")
    # Fetch data using the yfinance API
    data = yf.download(ticker_symbol, start=start_date, end=end_date)
    
    if data.empty:
        print("No data found. Check your ticker or date range.")
    else:
        print("Data successfully retrieved!")
        # Display the first 5 rows
        print(data.head())
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
output_dir = "collected_data"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# Save to CSV
file_path = os.path.join(output_dir, f"{ticker_symbol}_historical_data.csv")
data.to_csv(file_path)

print(f"Dataset saved successfully at: {file_path}")
```

```python
print(f"Total Records: {len(data)}")
print("\nMissing values per column:")
print(data.isnull().sum())

# Check data types
print("\nData Schema:")
print(data.dtypes)
```
