```python
import random as rd

#for random int using range
n=rd.randint(1,100)
print(n)
```

```python
#for random float without range
n=rd.random()
print(n)
```

```python
#for random float between range
n=rd.uniform(1.0,5.0)
print(n)
```

```python
#for random String between used
g=['Male','Female']
n=random.choice(g)
print(n)
```

```python
import pandas as pd
import random
record=[]
for i in range( 1,11):
    data={"roll_no":i,
          "marks":random.randint(1,100)
    }
    record.append(data)
df=pd.DataFrame(record)
print(df)
```
